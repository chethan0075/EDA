```python
import pandas as pd
import numpy as np

index=[('A',2010),('A',2020),('A',2025),('B',2010),('B',2020),('B',2025)]
population=[123,456,789,101,121,131]

i=pd.MultiIndex.from_tuples(index)
print(i)

pop=pd.Series(population)
pop=pop.reindex(index=i)
print(pop)

pop_df=pop.unstack()
print(pop_df)

pop_df.stack()

df=pd.DataFrame(np.random.rand(4,2),
                index=[['A','A','B','B'],[2010,2020,2010,2020]])
print(df)

data = {('A',2010):123, ('A',2020):456, ('A',2025):789, ('B',2020):1}
print(pd.Series(data))
```

```python
i1 = pd.MultiIndex.from_arrays(
    [['A','B','A'], [2010,2020,2025]],names=['country','year'])

pop1 = pd.Series(population[:3], index=i1)
print(pop1)
```

```python
i2=pd.MultiIndex.from_product([['A','B',],[2010,2020,2025]],names=["Country","Year"])
pop2=pd.Series(population,index=i2)
print(pop2)
```
