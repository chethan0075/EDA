```python
import numpy as np
arr = np.arange(1,21)
print(arr)
print(arr.ndim)
print(arr.shape)
print(arr.size)
print(arr.dtype)
```

```python
np.random.seed(1)
arr=np.random.randint(10,size=(4,5))
print(arr)
arr.nbytes
arr.itemsize
```

```python
arr=np.arange(1,13).reshape(3,4)
print(arr)
arr=np.arange(1,7).reshape(2,3)
print(arr)
```

```python
np.random.seed(1)
arr=np.random.randint(10,size=(3,4))
print(arr)
print(arr[:2,:2])
print(arr[:,1::2])
print(arr[::-1])
```
