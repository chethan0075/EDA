```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(1)

X = np.random.rand(10, 2)

print("Generated Points:")
print(X)

plt.scatter(X[:, 0], X[:, 1])
plt.xlabel("X-coordinate")
plt.ylabel("Y-coordinate")
plt.title("Random 2-D Points")
plt.grid(True)
plt.show()
```

```python
D = ((X[:, np.newaxis, :] - X[np.newaxis, :, :]) ** 2).sum(axis=2)

print("Distance Matrix:")
print(D)

X[:, np.newaxis, :]
X[np.newaxis, :, :]
```

```python
diff = X[:, np.newaxis, :] - X[np.newaxis, :, :]

print("Shape of coordinate differences:")
print(diff.shape)

sq_diff = diff ** 2

print("Shape of squared differences:")
print(sq_diff.shape)

D = sq_diff.sum(axis=2)
distance_matrix = sq_diff.sum(axis=2)

print("Shape of summed squared distances:")
print(distance_matrix.shape)

print("Summed squared distance matrix:")
print(distance_matrix)
```

```python
print("Diagonal elements:")
print(np.diag(D))

print("Are all diagonal elements zero?")
print(np.allclose(np.diag(D), 0))
```

```python
neighbors = np.argsort(D, axis=1)

print("Neighbor ordering:")
print(neighbors)

K = 2

nearest_indices = np.argpartition(D, K, axis=1)[:, :K+1]

print("K+1 nearest indices:")
print(nearest_indices)
```

```python
import numpy as np
import matplotlib.pyplot as plt

K = 2

nearest = np.argpartition(D, K, axis=1)[:, :K+1]

plt.figure(figsize=(8, 6))

plt.scatter(X[:, 0], X[:, 1], s=80)

for i in range(len(X)):
    plt.text(X[i, 0], X[i, 1], str(i))

for i in range(len(X)):
    for j in nearest[i, 1:]:
        plt.plot(
            [X[i, 0], X[j, 0]],
            [X[i, 1], X[j, 1]]
        )

plt.xlabel("X-coordinate")
plt.ylabel("Y-coordinate")
plt.title("2-Nearest Neighbor Network")
plt.grid(True)
plt.show()
```

```python
import numpy as np

np.random.seed(42)

X = np.random.rand(1000, 2)

D = ((X[:, np.newaxis, :] - X[np.newaxis, :, :]) ** 2).sum(axis=2)

K = 2

numpy_neighbors = np.argpartition(D, K, axis=1)[:, :K+1]

print("NumPy nearest neighbors for first 5 points:")
print(numpy_neighbors[:5])
```

```python
from sklearn.neighbors import KDTree

tree = KDTree(X)

distances, tree_neighbors = tree.query(X, k=3)

print("KD-Tree nearest neighbors for first 5 points:")
print(tree_neighbors[:5])
```

```python
numpy_sorted = np.take_along_axis(
    numpy_neighbors,
    np.argsort(
        np.take_along_axis(D, numpy_neighbors, axis=1),
        axis=1
    ),
    axis=1
)

print("NumPy:")
print(numpy_sorted[:5])

print("\nKD-Tree:")
print(tree_neighbors[:5])
```
